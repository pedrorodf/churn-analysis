# Python Insights - Analisando Dados com Python

### Case - Cancelamento de Clientes

```python
# Planejamento
#Passo 1: abrir a base de dados

import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")

#O ID não é relevante para a análise, então vamos removê-lo da tabela
tabela = tabela.drop(columns="CustomerID")

display(tabela)

#Passo 2: visualizar a base de dados
```

```python
# Passo 3: corrigir valores problemas da base de dados (tratar dados)

display(tabela.info())

# infos vazias
tabela = tabela.dropna()
```

```python
# Passo 4: análise inicial dos dados (descobrir padrões, tendências e insights)

#quantidade
display(tabela["cancelou"].value_counts())
#porcentagem
display(tabela["cancelou"].value_counts(normalize=True))

#porcentagem formatada
display(
    tabela["cancelou"]
    .value_counts(normalize=True)
    .map(lambda x: f"{x:.2%}"))
```

```python
# Passo 5: análise das causas dos cancelamentos (como as colunas da base impactam no cancelamento)
# gráficos/dashboards
# !pip install plotly
import plotly.express as px
import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")

# criar o grafico

for coluna in tabela.columns:
    grafico = px.histogram(tabela, x=coluna, color="cancelou")
    # exibir o grafico
    grafico.show()
```

```python
#Insights: 

# clientes do contrato mensal, onde TODOS cancelam
    # ofercer desconto nos planos anuais e trimestrais
# clientes que ligam mais do que 4 vezes para o call center, cancelam
    # criar um processo para resolver o problema do cliente em no máximo 3 ligações
# clientes que atrasaram mais de 20 dias, cancelaram
    # política de resolver atrasos em até 10 dias (equipe financeira)

tabela = tabela[tabela["duracao_contrato"]!="Monthly"]
tabela = tabela[tabela["ligacoes_callcenter"]<=4]
tabela = tabela[tabela["dias_atraso"]<=20]

display(tabela["cancelou"].value_counts())
# em percentual
display(tabela["cancelou"].value_counts(normalize=True))
```
